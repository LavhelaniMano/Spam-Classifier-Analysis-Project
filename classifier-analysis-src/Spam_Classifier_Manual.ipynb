```python
import pandas as pd # For data manipulation (think of it as Excel on steroids)
from sklearn.model_selection import train_test_split # To split our data into study material and test material
from sklearn.feature_extraction.text import CountVectorizer # To turn words into numbers (math!)
from sklearn.naive_bayes import MultinomialNB # The algorithm we will use to train
from sklearn.metrics import accuracy_score, classification_report # To grade our AI's performance

print("Toolkits loaded and ready to go!")
```

```python
# We create a dictionary with our expanded text messages and their labels
data = {
    'text': [
        # --- HAM (Real Messages) - 15 Examples ---
        "Hey, are we still on for the bootcamp tomorrow?",
        "Can you send me the Python curriculum notes?",
        "Just finished the GraphRAG project, it looks amazing!",
        "Let's grab coffee and review the pitch deck.",
        "Did you push the latest code to GitHub?",
        "Mom, can you pick up some milk on your way home?",
        "The quarterly financial report is attached for your review.",
        "Are we still playing basketball this weekend?",
        "Meeting moved to 3 PM in the main conference room.",
        "Happy anniversary! Love you so much.",
        "Please approve the PTO request I submitted yesterday.",
        "The Wi-Fi password for the guest network is on the fridge.",
        "Did you see the game last night? What a finish!",
        "I'll be about 10 minutes late to the dinner reservation.",
        "Can you forward me the slides from yesterday's presentation?",

        # --- SPAM (Junk Messages) - 15 Examples ---
        "WINNER!! Claim your free iPhone 15 now! Click here",
        "URGENT: Your bank account has been locked. Reply to unlock.",
        "Congratulations! You've been selected for a $1000 gift card.",
        "Double your crypto in 24 hours!!! Send Bitcoin now.",
        "HOT DEAL: Get 90% off designer watches! Buy now!",
        "ACT NOW! Your car warranty is about to expire. Call us!",
        "You have a pending package delivery. Track it here: bit.ly/track",
        "Earn $5000 a week from home! No experience required. Apply now.",
        "FINAL NOTICE: Your Netflix subscription has failed. Update billing.",
        "Singles in your area are waiting to meet you! Click to chat.",
        "Congratulations! You are the 1,000,000th visitor. Claim your prize!",
        "Your Apple ID has been locked for suspicious activity. Verify now.",
        "Lose 20 pounds in 2 weeks with this one weird trick!",
        "URGENT: Wire transfer of $5,000 needed immediately to release funds.",
        "Free vacation to the Bahamas! Just pay the $50 processing fee."
    ],
    # Notice we have exactly 30 labels to match the 30 text messages above
    'label': [
        'ham', 'ham', 'ham', 'ham', 'ham', 'ham', 'ham', 'ham', 'ham', 'ham', 'ham', 'ham', 'ham', 'ham', 'ham',
        'spam', 'spam', 'spam', 'spam', 'spam', 'spam', 'spam', 'spam', 'spam', 'spam', 'spam', 'spam', 'spam', 'spam', 'spam'
    ]
}

# Convert it into a Pandas DataFrame (a neat table)
df = pd.DataFrame(data)

# Let's peek at our data!
print("Here is the data our AI will learn from:\n")
print(df)
```

```python
import pandas as pd

# 1. Split the data!
# X = the text messages, y = the answers (spam/ham)
# test_size=0.3 means 30% of data is hidden for testing later.
# You will frequently see random_state=42. The number 42 has no special mathematical power; it is simply a popular sci-fi joke from the book The Hitchhiker's Guide to the Galaxy. You can use any integer like 0, 1, or 123 to achieve the exact same effect.
# Added stratify=df['label'] to ensure an even distribution of 'ham' and 'spam' in both train and test sets.
X_train, X_test, y_train, y_test = train_test_split(df['text'], df['label'], test_size=0.5, random_state=42, stratify=df['label'])

# 2. Initialize the Word Counter (Vectorizer)
vectorizer = CountVectorizer()

# 3. Translate the training text into a mathematical matrix
X_train_math = vectorizer.fit_transform(X_train)

# 4. Translate the testing text using the SAME vocabulary
X_test_math = vectorizer.transform(X_test)

print("Words successfully converted to numbers!")
```

```python
# Initialize the Naive Bayes AI model
ai_classifier = MultinomialNB()

# Train the model! The .fit() command is where the actual learning happens.
# We give it the math-text (X_train_math) and the answers (y_train).
ai_classifier.fit(X_train_math, y_train)

print("The AI has successfully trained itself!")
```

```python
# Ask the AI to predict the answers for our hidden test set
predictions = ai_classifier.predict(X_test_math)

# Grade the test! Compare the AI's predictions to the actual real answers
accuracy = accuracy_score(y_test, predictions)

print(f"AI Accuracy Score: {accuracy * 100}%\n")

print("Detailed Report Card:")
print(classification_report(y_test, predictions))
```

```python
def check_spam(message):
    # 1. Turn the new message into math
    math_message = vectorizer.transform([message])

    # 2. Get the AI's prediction
    prediction = ai_classifier.predict(math_message)[0]

    # 3. Print the result with some flair
    if prediction == 'spam':
        print(f"SPAM DETECTED: '{message}'")
    else:
        print(f"SAFE (Ham): '{message}'")
```

```python
# TRY IT YOURSELF!
# Change the text inside the quotes to trick the AI!

check_spam("Can you send me the meeting notes?")
check_spam("Are we still meeting at 2 PM?")
check_spam("Please review my assignment before tomorrow.")

check_spam("Congratulations! You have won $10,000! Click here now!")
check_spam("FREE iPhone! Claim your prize immediately!")
check_spam("URGENT: Your account has been suspended. Verify your details now!")

check_spam("Limited-time discount on all products!")
check_spam("Exclusive offer just for you.")
check_spam("Don't miss out on this amazing deal!")
```
